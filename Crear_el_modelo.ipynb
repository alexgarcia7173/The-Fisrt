```python
# Librerias para hacer una red neuronal con tensor flow y Keras
import tensorflow as tf
from tensorflow.keras import layers, models # crear capas y modelos en red
from tensorflow.keras.datasets import mnist #importa el dataset mnist
import numpy as np
```

cargar conjunto de datos

```python
# Paso 2: Cargar el conjunto de datos MNIST
(x_train, y_train), (x_test, y_test) = mnist.load_data()
#mnist.load_data() funcion que carga el dataset en forma de array
# x_train contiene las imagenes para entrenar el modelo
#y_train contiene etiquetas:digito correcto de la imagen
#x_test imagenes para evaluar el modelo entrenado
#y_test etiquetas para los datso de prueba
```

Procesar los datos

```python
# Paso 3: Preprocesar los datos
x_train = x_train.astype('float32') / 255  # Normalización
x_test = x_test.astype('float32') / 255
#astype('float32')convierte la imagen a float
#/ 255 divide cada pixel para que queden valores entre 0 y 1
# Aplanar las imágenes
#trasnformar las imagenes de unarray 2 D(28*28) a un array 1D
x_train = x_train.reshape(-1, 28 * 28)
x_test = x_test.reshape(-1, 28 * 28)
```

```python
# One-hot encoding
#tf.keras.utils.to_categorical funcion que conviete las etiquetas
# a formae one-hot-encoding
#y_train, y_test contienen valores enteros que reprsentan los digitos
# de cada imagen
# argumento 10 especifica el numeto de categorias en el dataset
y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test = tf.keras.utils.to_categorical(y_test, 10)
```

crear el modelo

```python
# Paso 4: Definir el modelo MLP (perceptron multicapas)
model = models.Sequential() # incia modelo secuencial vacio
# model.add(layer) adiciona la primera capa
# larer.dense significa que las neuronas estan conectada con la capa
#previa
# 512 es el numero de neuronas
# relu funcion de acivacion introduce no linearidad para identificar
#patrones complejos
# input_shaoe 28 es la forma de ingreso de los datos un array 1D de 28*28
model.add(layers.Dense(512, activation='relu', input_shape=(28 * 28,)))
model.add(layers.Dense(256, activation='relu'))
model.add(layers.Dense(128, activation='relu'))
model.add(layers.Dense(10, activation='softmax'))
# model.add(layers.Dense(10, activation='softmax')) ultima capa
# 10 neuronas corresponde a los 10 posibles digios de salida
# sofmax convierte la salida de la red en probabilidades para cada clase
```

Compilar el modelo

```python
# Paso 5: Compilar el modelo es decir configurarlo para entrenamiento
# reglas que el modelo usara para aprender
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
#optimizer='adam', el algoritmo usara los pesos y sesgos internos
# loss= define la funcion que se va a usar para evalaur las diferencias
#entre las prediciones y los valores target
#'categorical_crossentropy' funcion de perdida para problemas de clasificacion
# accuracy metrica de desempeño, calcula el % de ejemplos correctamente
#clasificados
```

Entrenar el modelo

```python
# Paso 6: Entrenar el modelo
history = model.fit(x_train, y_train, epochs=10, batch_size=128, validation_split=0.2)
#history variable que guarda el la salida de fit()
#x_train, y_train son los datos de entrenamiento y sus etiquetas
# epochs 10 el modelo itera 10 veces en el dataset de entrenamiento
#batch_size=128 durante cada iteracion los datos de entranamiento se
#dividen lotes en este caso de 128 muestras
# el modelo actualiza los pesos despues de procesar cada lote
#validation_split=0.2 especififica que el 20 % de los datos de
#entrenamiento se utizan para validacion
```

Evaluar el modelo

```python
# Paso 7: Evaluar el modelo
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f'Precisión en el conjunto de prueba: {test_acc}')
# funcion evaluate()compara las prediciones del modelo en x_test con
#las etiquetas actules en y_test
#test_loss que tanto las prediciones se acoplan a las verdaderas etiquetas
# entre menos es mejor
#Test_accprecision del modelo,porcentaje de imagenes correctamente clasificadas
```

```python
import streamlit as st
from PIL import Image
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model
#PIL libreria de imagnes en Pythos tiene herramientas para trabajar
# com imagenes
#tensorflow.keras.models import load_model prmite cargar modelos guardados
```

```python
# cargar el modelo entrenado
model = load_model('my_model.keras')
```

```python
#Función para preprocesar la imagen
def preprocess_image(image):
    image = image.convert('L')  # Convertir a escala de grises
    image = image.resize((28, 28))  # Redimensionar a 28x28 píxeles
    image = np.array(image) #convertir a array numpy
    image = image / 255.0 #Normalizar valores entre 0 y 1
    image = np.reshape(image,(1, 28*28)) #Redimensionar para la entrada del modelo (1= representa el 3 de imagen)
                                                                                            #(28, 28= pixeles de sncho y alto)
                                                                                            #(1 significa escala de grises, escala rgb se coloca un 3)
    return image
```
